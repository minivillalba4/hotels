Casteo de Variables

```python
df_raw2 = df_raw.copy()  # Crear una copia del DataFrame original para trabajar sin modificarlo
```

```python

# 1. Fechas (object -> datetime)
df_raw2['reservation_status_date'] = pd.to_datetime(df_raw2['reservation_status_date'], errors='coerce')
```

```python
# 2. Booleanos (int64 -> bool)
df_raw2['is_canceled']       = df_raw2['is_canceled'].astype('bool')
df_raw2['is_repeated_guest'] = df_raw2['is_repeated_guest'].astype('bool')
```

```python
# 3. Categóricas nominales (object -> category)
cat_cols = ['hotel', 'meal', 'country', 'market_segment',
            'distribution_channel', 'reserved_room_type',
            'assigned_room_type', 'deposit_type',
            'customer_type', 'reservation_status']
df_raw2[cat_cols] = df_raw2[cat_cols].astype('category')
```

```python
# 4. Categórica ordenada (meses)
orden_meses = CategoricalDtype(
    categories=['January', 'February', 'March', 'April', 'May', 'June',
                'July', 'August', 'September', 'October', 'November', 'December'],
    ordered=True)
df_raw2['arrival_date_month'] = df_raw2['arrival_date_month'].astype(orden_meses)
```

```python
# 5. Entero nullable (float64 con NaN -> Int64)
df_raw2['children'] = df_raw2['children'].astype('Int64')
```

```python
# 6. IDs que llegan como float -> rellenar nulos y pasar a category
df_raw2['agent']   = df_raw2['agent'].fillna(0).astype('int64').astype('category')
df_raw2['company'] = df_raw2['company'].fillna(0).astype('int64').astype('category')
```

```python
# 7. Nulos en country (488) -> "Unknown"
df_raw2['country'] = df_raw2['country'].cat.add_categories('desconocido').fillna('desconocido')
```

## Tratamiento para EDA (corte duro)

Criterio: umbrales fijos elegidos a ojo sobre las distribuciones para mejorar la visualización del EDA. **No aplicable a ML** — en ML se entrenará una técnica propia (IQR, percentil, etc.) sobre el raw para que pueda reaplicarse a datos nuevos.

- **Duplicados**: no se tocan. El dataset no tiene columna ID, así que no podemos distinguir duplicación real de reservas legítimamente idénticas (p.ej. bloques de grupo).
- **Nulos**:
  - Categóricas (`country`, `agent`, `company`): ya tratadas arriba con etiqueta explícita.
  - Numéricas (`children`, 4 nulos): imputación por la mediana (=0).
- **Atípicos**: cortes duros por variable (ver celda siguiente). Pérdida esperada ≲ 1,5 % del dataset, concentrada en colas claramente patológicas (reservas sin huéspedes, ADR de 5.400 €, 55 adultos en una habitación, etc.).

```python
import logging
logging.basicConfig(level=logging.INFO, format='%(message)s', force=True)
log = logging.getLogger(__name__)

filas_inicial = len(df_raw2)

# 1) Nulos numéricos residuales -> mediana (children: 4 nulos, mediana = 0)
df_raw2['children'] = df_raw2['children'].fillna(df_raw2['children'].median()).astype('Int64')

# 2) Cortes duros (umbrales fijos NO reentrenables, solo para EDA)
#    En ML se aplicará otra estrategia (IQR / percentil entrenable) sobre el raw.
cortes = {
    'lead_time':                      lambda s: s <= 500,
    'stays_in_weekend_nights':        lambda s: s <= 6,
    'stays_in_week_nights':           lambda s: s <= 14,
    'days_in_waiting_list':           lambda s: s <= 120,
    'booking_changes':                lambda s: s <= 5,
    'adults':                         lambda s: s.between(1, 4),
    'children':                       lambda s: s.between(0, 3),
    'babies':                         lambda s: s <= 2,
    'required_car_parking_spaces':    lambda s: s <= 3,
    'total_of_special_requests':      lambda s: s <= 5,
    'previous_cancellations':         lambda s: s <= 5,
    'previous_bookings_not_canceled': lambda s: s <= 20,
    'adr':                            lambda s: s.between(0, 450),
}

log.info("Cortes duros aplicados:")
for col, regla in cortes.items():
    n_prev = len(df_raw2)
    df_raw2 = df_raw2[regla(df_raw2[col])]
    eliminadas = n_prev - len(df_raw2)
    if eliminadas:
        log.info(f"  {col:<34s} -> -{eliminadas:>5,}")

df_raw2 = df_raw2.reset_index(drop=True)

# 3) Resumen
log.info("")
log.info(f"Filas inicio : {filas_inicial:>7,}")
log.info(f"Filas final  : {len(df_raw2):>7,}  ({len(df_raw2)/filas_inicial:.2%} retenido)")
log.info(f"Nulos totales: {df_raw2.isnull().sum().sum()}")
```
